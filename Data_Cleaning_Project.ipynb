```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("archive (2)/my_file (1).csv")

df.head()
```

```python
print("Peak column:")
print(df["Peak"].unique())

print("\nAll Time Peak column:")
print(df["All Time Peak"].unique())
```

```python
df["Peak"] = df["Peak"].fillna("Not Available")
df["All Time Peak"] = df["All Time Peak"].fillna("Not Available")

df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.columns = df.columns.str.replace("\xa0", " ", regex=False)
df.columns = df.columns.str.strip()

print(df.columns)
```

```python
print(df["Actual gross"].head())
print(df["Adjusted gross (in 2022 dollars)"].head())
print(df["Average gross"].head())
```

```python
gross_columns = [
    "Actual gross",
    "Adjusted gross (in 2022 dollars)",
    "Average gross"
]

for column in gross_columns:
    df[column] = (
        df[column]
        .astype(str)
        .str.replace("$", "", regex=False)
        .str.replace(",", "", regex=False)
        .str.strip()
    )
    
    df[column] = pd.to_numeric(df[column], errors="coerce")

df[gross_columns].head()
```

```python
df[gross_columns].isnull().sum()
```

```python
df[df["Actual gross"].isnull()]
```

```python
median_gross = df["Actual gross"].median()

df["Actual gross"] = df["Actual gross"].fillna(median_gross)

df["Actual gross"].isnull().sum()
```

```python
df["Peak"] = df["Peak"].astype(str).str.extract(r"(\d+)")[0]
df["All Time Peak"] = df["All Time Peak"].astype(str).str.extract(r"(\d+)")[0]

print(df[["Peak", "All Time Peak"]].head(10))
```

```python
df["Peak"] = df["Peak"].fillna("Not Available")
df["All Time Peak"] = df["All Time Peak"].fillna("Not Available")

df[["Peak", "All Time Peak"]].head(10)
```

```python
df["Year(s)"].unique()
```

```python
df["Start Year"] = df["Year(s)"].str.extract(r"(\d{4})")[0]

df["End Year"] = df["Year(s)"].str.extract(
    r"–(\d{4})"
)[0]

# If there is no end year, use the start year
df["End Year"] = df["End Year"].fillna(df["Start Year"])

# Convert both columns to numeric
df["Start Year"] = df["Start Year"].astype(int)
df["End Year"] = df["End Year"].astype(int)

df[["Year(s)", "Start Year", "End Year"]].head(10)
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
numeric_columns = [
    "Actual gross",
    "Adjusted gross (in 2022 dollars)",
    "Shows",
    "Average gross"
]

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]

    print(f"\n{column}: {len(outliers)} possible outlier(s)")
```

### Outlier Analysis

The IQR method was used to identify possible outliers in the numerical columns. Some possible outliers were detected in revenue and show-related variables.

These values were not removed because the dataset represents real concert tour performance, and unusually high revenue or a high number of shows may represent genuine successful tours rather than data errors.
