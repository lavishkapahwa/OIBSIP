Before performing analysis, the dataset was checked for missing values,

# 5. Exploratory Data Analysis
The cleaned dataset is now analyzed to identify important patterns,
trends, relationships, and business insights related to sales, profit,
products, customers, regions, and shipping.

```python
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()

print("Total Sales:", total_sales)
print("Total Profit:", total_profit)
```

```python
average_sales = df["Sales"].mean()
average_profit = df["Profit"].mean()

print("Average Sales:", average_sales)
print("Average Profit:", average_profit)
```

```python
category_profit = df.groupby("Product Category")["Profit"].sum()

print(category_profit)
```

```python
category_sales = df.groupby("Product Category")["Sales"].sum()

print(category_sales)
```

```python
region_sales = df.groupby("Region")["Sales"].sum()

print(region_sales)
```

```python
region_profit = df.groupby("Region")["Profit"].sum()

print(region_profit)
```

```python
category_quantity = df.groupby("Product Category")["Order Quantity"].sum()

print(category_quantity)
```

```python
subcategory_profit = df.groupby("Product Sub-Category")["Profit"].sum()

print(subcategory_profit.sort_values(ascending=False))
```

```python
top_products = df.groupby("Product Name")["Sales"].sum()

print(top_products.sort_values(ascending=False).head(10))
```

```python
top_profit_products = df.groupby("Product Name")["Profit"].sum()

print(top_profit_products.sort_values(ascending=False).head(10))
```

```python
segment_sales = df.groupby("Customer Segment")["Sales"].sum()

print(segment_sales)
```

```python
segment_profit = df.groupby("Customer Segment")["Profit"].sum()

print(segment_profit)
```

```python
yearly_sales = df.groupby(df["Order Date"].dt.year)["Sales"].sum()

print(yearly_sales)
```

```python
yearly_profit = df.groupby(df["Order Date"].dt.year)["Profit"].sum()

print(yearly_profit)
```

```python
monthly_sales = df.groupby(df["Order Date"].dt.month)["Sales"].sum()

print(monthly_sales)
```

```python
monthly_profit = df.groupby(df["Order Date"].dt.month)["Profit"].sum()

print(monthly_profit)
```

```python
ship_sales = df.groupby("Ship Mode")["Sales"].sum()

print(ship_sales)
```

```python
ship_profit = df.groupby("Ship Mode")["Profit"].sum()

print(ship_profit)
```

```python
priority_sales = df.groupby("Order Priority")["Sales"].sum()

print(priority_sales.sort_values(ascending=False))
```

```python
priority_profit = df.groupby("Order Priority")["Profit"].sum()

print(priority_profit.sort_values(ascending=False))
```

```python
category_discount = df.groupby("Product Category")["Discount"].mean()

print(category_discount)
```

```python
shipping_cost = df.groupby("Ship Mode")["Shipping Cost"].sum()

print(shipping_cost.sort_values(ascending=False))
```

```python
correlation = df[[
    "Sales",
    "Profit",
    "Discount",
    "Order Quantity",
    "Unit Price",
    "Shipping Cost"
]].corr()

print(correlation)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.scatter(df["Sales"], df["Profit"])
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.title("Sales vs Profit")
plt.show()
```

### Key Insights

- Technology generated the highest sales and profit.
- Central region recorded the highest sales and profit.
- Corporate customers contributed the most revenue and profit.
- Regular Air was the most profitable shipping mode.
- December had the highest sales, while October had the highest profit.
- Sales and Profit showed a moderate positive relationship.

# 6. Conclusion

The exploratory data analysis of the Superstore Sales dataset provided useful
insights into sales, profit, products, customers, regions, and shipping methods.

Technology generated the highest sales and profit, while the Central region
recorded the highest overall sales and profit. Corporate customers contributed
the highest sales and profit among the customer segments. Regular Air was the
most successful shipping mode based on sales and profit.

The analysis also showed a moderate positive relationship between Sales and
Profit. Overall, the EDA helped identify important business patterns and
provided insights that can support better business decisions.
